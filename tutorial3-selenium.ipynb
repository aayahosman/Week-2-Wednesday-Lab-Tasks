### Locate Elements in the DOM

````python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By

options = Options()
options.add_argument("--headless")

driver = webdriver.Chrome(options=options)
driver.implicitly_wait(5)

driver.get("https://bandcamp.com/discover/")

print(driver.title)

pagination_button = driver.find_element(By.ID, "view-more")

print(pagination_button.accessible_name)

tracks = driver.find_elements(
    By.CLASS_NAME,
    "results-grid-item"
)

print(len(tracks))
print(tracks[0].text)

driver.quit()
````

## Interact With Web Elements

### Click Buttons and Links

````python
import time

from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.common.exceptions import ElementNotInteractableException

options = Options()
options.add_argument("--headless")

driver = webdriver.Chrome(options=options)

driver.get("https://bandcamp.com/discover/")

pagination_button = driver.find_element(By.ID, "view-more")

try:
    pagination_button.click()
except ElementNotInteractableException:
    print("Button could not be clicked.")

time.sleep(1)

driver.quit()
````

### Send Keystrokes and Text Entry

````python
import time
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.common.exceptions import NoSuchElementException
from selenium.webdriver.common.by import By

options = Options()
options.add_argument("--headless")

driver = webdriver.Chrome(options=options)
driver.implicitly_wait(5)

driver.get("https://bandcamp.com/discover/")

try:
    cookie_accept_button = driver.find_element(
        By.CSS_SELECTOR,
        "#cookie-control-dialog button.g-button.outline"
    )
    cookie_accept_button.click()

except NoSuchElementException:
    pass

time.sleep(0.5)

search = driver.find_element(
    By.CLASS_NAME,
    "site-search-form"
)

search_field = search.find_element(
    By.TAG_NAME,
    "input"
)

search_field.send_keys("selenium")
search_field.submit()

time.sleep(1)

driver.quit()
````

### Deal With Hidden or Overlaid Elements

The cookie consent pop-up can cover elements on the page. I used a `try`/`except` statement to click the cookie button if it appears. If it does not appear, `NoSuchElementException` is ignored and the program continues.


### Use Hover, Drag-and-Drop, and More Complex Gestures

Selenium can use `ActionChains` for more complex actions like hovering and clicking.

```python
from selenium.webdriver import ActionChains

menu = driver.find_element(By.CSS_SELECTOR, ".menu")
submenu = driver.find_element(By.CSS_SELECTOR, ".menu #submenu")

actions = ActionChains(driver)
actions.move_to_element(menu)
actions.click(submenu)
actions.perform()
```

### Submit Forms

Selenium can fill out multiple fields and then submit a form.

```python
signup_form = driver.find_element(By.ID, "signup-form")

email_input = signup_form.find_element(By.NAME, "email")
password_input = signup_form.find_element(By.NAME, "password")

email_input.send_keys("user@example.com")
password_input.send_keys("MySecurePassword123")

signup_form.submit()
```

### Understand Implicit, Explicit, and Fluent Waits

### Set Up an Implicit Wait

An implicit wait tells Selenium to keep checking for an element for a certain amount of time before giving up.

```python
driver.implicitly_wait(5)
```

### Use Explicit Waits for Targeted Waiting

An explicit wait waits for a specific condition before continuing.

```python
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

wait = WebDriverWait(driver, 10)

wait.until(
    EC.element_to_be_clickable(
        (By.ID, "view-more")
    )
)
```
